## Import libraries and packages

```python
import os
import geopandas as gpd
import rasterio
from rasterio.warp import calculate_default_transform, reproject, Resampling
```

## Prepare paths con convert GIS Data to WGS 1984

```python
# Ruta del shapefile original
shp_entrada = "ASB_Caribe_shape.shp"
shp_salida = "ASB_Caribe_4326.shp"

# Leer el archivo vectorial
gdf = gpd.read_file(shp_entrada)

print(f"Sistema de coordenadas original: {gdf.crs}")

# Reproyectar a WGS 84 (EPSG:4326)
gdf_4326 = gdf.to_crs(epsg=4326)

# Guardar el nuevo Shapefile
gdf_4326.to_file(shp_salida)
print("¡Shapefile reproyectado a EPSG:4326 y guardado con éxito!")
```
